```python
import tiktoken
```

```python
tokenizer = tiktoken.encoding_for_model("gpt-4o")
```

```python
prompt = "Vous êtes un expert en IA agentique. C'est un agent AI"
```

```python
tokens = tokenizer.encode(prompt)
print(tokens)
print(f"Le prompt contient {len(tokens)} tokens.")
```

```python
for token in tokens:
    t = tokenizer.decode_single_token_bytes(token).decode("utf-8")
    print(f"Token: {token}, Caractère: '{t}'")
```

```python
from langchain_openai import ChatOpenAI
from dotenv.ipython import load_dotenv
load_dotenv()
```

```python
llm = ChatOpenAI(model="gpt-4o", temperature=0)
```

```python
from IPython.display import Markdown, display
```

```python
response = llm.invoke(input = [
    {"role": "system", "content": "Vous êtes un expert en IA Agentique"},
    {"role": "user", "content": "Qu'est-ce que un Agent AI?"}
])
print(display(Markdown(response.content)))
```

```python
llm3 = ChatOpenAI(model="gpt-5.2", temperature=0.5)
llm_with_tools = llm3.bind_tools([
    {   "type": "image_generation","quality": "high"}
])
```
